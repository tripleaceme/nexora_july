```python
# validate the date values in the birthdate column of the patients dataframe
for index, row in df_patients_cl.iterrows():
    try:
        pd.to_datetime(row["BIRTHDATE"])
    except ValueError:
        print(f"Row {index}: Invalid date value in the BirthDate column: {row['BIRTHDATE']}")
```

```python
# Convert the "BIRTHDATE" column to datetime format, handling errors by coercing invalid formats to NaT (Not a Time).
df_patients_cl["BIRTHDATE"] = pd.to_datetime(df_patients_cl["BIRTHDATE"])
```

```python
# convert deathdate to datetime format, handling errors by coercing invalid formats to NaT (Not a Time).
df_patients_cl["DEATHDATE"] = pd.to_datetime(df_patients_cl["DEATHDATE"], errors='coerce')
```

```python
# drop ssn, drivers, passpor and prefix columns from the dataframe as they are not needed for further analysis.
df_patients_cl.drop(columns=["SSN", "DRIVERS", "PASSPORT", "PREFIX"], inplace=True)
```

```python
# fill missing values in marital with unknown as it is a categorical variable and we don't want to lose any data.
df_patients_cl["MARITAL"].fillna("Unknown", inplace=True)
```

```python
# strip off the numbers in the first, middle and last name columns to retain only the alphabetic characters.
df_patients_cl["FIRST"] = df_patients_cl["FIRST"].str.replace(r'\d+', '', regex=True)
df_patients_cl["LAST"] = df_patients_cl["LAST"].str.replace(r'\d+', '', regex=True)
```

```python
# combine first, middle and last name columns into a single column called "FULL_NAME"
df_patients_cl["FULL_NAME"] = df_patients_cl[["FIRST", "MIDDLE", "LAST"]].apply(lambda x: ' '.join(x.dropna()), axis=1)
```

```python
# drop first, middle and last name columns as they are no longer needed after creating the FULL_NAME column.
# drop suffix and maiden name columns as they are not needed for further analysis.
df_patients_cl.drop(columns=["FIRST", "MIDDLE", "LAST", "SUFFIX", "MAIDEN"], inplace=True)
```

```python
# Replace all missing values of Marital status with unknown as it is a categorical variable and we don't want to lose any data.
df_patients_cl["MARITAL"] = df_patients_cl["MARITAL"].map({"M": "Married", "S": "Single",
                                                            "D": "Divorced", "W": "Widowed","Unknown": "Unknown"})
```

```python
# Standardize the "RACE", "ETHNICITY" columns by converting all values to title case (first letter capitalized) for consistency.
df_patients_cl["RACE"] = df_patients_cl["RACE"].str.title()
df_patients_cl["ETHNICITY"] = df_patients_cl["ETHNICITY"].str.title()
```

```python
# Map the "GENDER" column values from "M" and "F" to "Male" and "Female" respectively for clarity and consistency
df_patients_cl["GENDER"] = df_patients_cl["GENDER"].map({"M": "Male", "F": "Female"})
```

```python
# update the birthplace for patient with the Id c93095b90340 to Shanghai  
# Shanghai-Municipality  CN so as to have a consistent format for all the birthplace values.
df_patients_cl.loc[df_patients_cl["Id"] == "c93095b90340", "BIRTHPLACE"] = "Shanghai  Shanghai-Municipality  CN" 
```

```python
df_patients_cl.head(2)
```

```python
# split the "BIRTHPLACE" column into three separate columns: "BIRTHPLACE_CITY", "BIRTHPLACE_STATE", 
# and "BIRTHPLACE_COUNTRY" by splitting the string on spaces and extracting the respective parts.

df_patients_cl["BIRTHPLACE_CITY"] = df_patients_cl["BIRTHPLACE"].str.split(" ").str[0]
df_patients_cl["BIRTHPLACE_STATE"] = df_patients_cl["BIRTHPLACE"].str.split(" ").str[2]
df_patients_cl["BIRTHPLACE_COUNTRY"] = df_patients_cl["BIRTHPLACE"].str.split(" ").str[4]
```

```python
df_patients_cl.head(2)
```

```python
# drop birthplace column as it is no longer needed after creating the BIRTHPLACE_CITY, BIRTHPLACE_STATE and BIRTHPLACE_COUNTRY columns.
df_patients_cl.drop(columns=["BIRTHPLACE"], inplace=True)

# Rename resident to the city, state and country columns name to have a consistent format for all the birthplace values.
df_patients_cl.rename(columns={"CITY": "RESIDENT_CITY", 
                               "STATE": "RESIDENT_STATE", 
                               "COUNTY": "RESIDENT_COUNTY"}, inplace=True)
```

```python
# Convert the "FIPS" column to string type to ensure consistency in data type for further analysis.
df_patients_cl['FIPS'] = df_patients_cl['FIPS'].astype(str)
```

```python
# Read the missing FIPS codes from the CSV file into a DataFrame called fill_fips.
fill_fips = pd.read_csv(root_path / "missing_fips.csv")
```

```python
fill_fips.head()
```

```python
# Merge the df_patients_cl DataFrame with the fill_fips DataFrame on the "RESIDENT_COUNTY" column from df_patients_cl and the "county" column from fill_fips.
# For testing purposes, we will only merge the first 10 rows of df_patients_cl with fill_fips to check if the merge is working correctly.
missing_fips = df_patients_cl[["RESIDENT_STATE", "RESIDENT_COUNTY", "FIPS"]]
```

```python
missing_fips
```

```python
# Merge the missing_fips DataFrame with the fill_fips DataFrame on the "RESIDENT_COUNTY" column from missing_fips and the "county" column from fill_fips.
df_fips = missing_fips.merge(
    fill_fips,
    how='left',
    left_on='RESIDENT_COUNTY',
    right_on='county'
)
```

```python
df_fips
```

```python
# Merge the df_patients_cl DataFrame with the fill_fips DataFrame on the "RESIDENT_COUNTY" column from df_patients_cl and the "county" column from fill_fips.
df_patients_cl = df_patients_cl.merge(
    fill_fips,
    how='left',
    left_on='RESIDENT_COUNTY',
    right_on='county'
)
```

```python
df_patients_cl.head(2)
```

```python
# drop county column as it is no longer needed after merging with the fill_fips dataframe.
df_patients_cl.drop(columns=["county"], inplace=True)
```

```python
df_patients_cl.head(2)
```

```python
# drop the .0 from the FIPS column values to ensure consistency in the format of the FIPS codes.
df_patients_cl["FIPS"] = df_patients_cl["FIPS"].str.replace(r'\.0$', '', regex=True)
```
